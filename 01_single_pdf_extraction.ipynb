```python
import fitz
from pathlib import Path

# Path to your PDF
pdf_path = Path(r"C:\Users\Tanvi Singh\OneDrive\Desktop\articles\Baloch_2022.pdf")

# Open PDF
doc = fitz.open(pdf_path)

full_text = ""

# Show total pages
print("Total pages:", len(doc))

# Extract text page by page
for page_num, page in enumerate(doc, start=1):
    text = page.get_text()

    full_text += f"\n\n--- PAGE {page_num} ---\n"
    full_text += text

# Print first 3000 characters
print(full_text)
```

```python
print(len(full_text))
```

```python
from pathlib import Path

output_path = Path(r"C:\Users\Tanvi Singh\OneDrive\Desktop\articles\Baloch_2022.pdf")

output_path.write_text(full_text, encoding="utf-8")

print("Text saved successfully")
```

```python
print(output_path)
print(output_path.exists())
```
