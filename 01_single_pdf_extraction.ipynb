```python
from pydantic import BaseModel
from typing import Optional

class EvidenceFinding(BaseModel):
    study_title: Optional[str] = None
    year: Optional[int] = None
    population: Optional[str] = None
    sample_size: Optional[int] = None
    sepsis_type: Optional[str] = None
    biomarker: Optional[str] = None
    outcome: Optional[str] = None
    main_finding: Optional[str] = None
    odds_ratio: Optional[float] = None
    auc: Optional[float] = None
    p_value: Optional[str] = None
    source_quote: Optional[str] = None
    chunk_id: Optional[int] = None
```

```python
import fitz
import json
import re
from pathlib import Path

pdf_folder = Path(r"C:\Users\Tanvi Singh\OneDrive\Desktop\articles")

for pdf_file in pdf_folder.glob("*.pdf"):

    print(f"\nProcessing: {pdf_file.name}")

    try:
        doc = fitz.open(pdf_file)
    except Exception as e:
        print(f"Skipping {pdf_file.name} because it could not be opened.")
        print("Error:", e)
        continue

    full_text = ""

    for page_num, page in enumerate(doc, start=1):
        text = page.get_text()
        full_text += f"\n\n--- PAGE {page_num} ---\n{text}"

    extracted_path = pdf_folder / f"{pdf_file.stem}_extracted.txt"
    extracted_path.write_text(full_text, encoding="utf-8")

    clean_text = re.sub(r'--- PAGE \d+ ---', '', full_text)
    clean_text = clean_text.replace("\n", " ")
    clean_text = " ".join(clean_text.split())

    cleaned_path = pdf_folder / f"{pdf_file.stem}_cleaned.txt"
    cleaned_path.write_text(clean_text, encoding="utf-8")

    chunk_size = 2000
    chunks = [
        clean_text[i:i + chunk_size]
        for i in range(0, len(clean_text), chunk_size)
    ]

    chunk_path = pdf_folder / f"{pdf_file.stem}_chunks.json"

    with open(chunk_path, "w", encoding="utf-8") as f:
        json.dump(chunks, f, indent=2)

    print(f"Finished: {pdf_file.name}")
```

```python
import json
from pathlib import Path

folder = Path(r"C:\Users\Tanvi Singh\OneDrive\Desktop\articles")

all_chunks = []

for chunk_file in folder.glob("*_chunks.json"):
    print("Reading:", chunk_file.name)

    with open(chunk_file, "r", encoding="utf-8") as f:
        chunks = json.load(f)

    paper_name = chunk_file.name.replace("_chunks.json", "")

    for idx, chunk_text in enumerate(chunks):
        all_chunks.append({
            "paper_name": paper_name,
            "chunk_id": idx,
            "text": chunk_text
        })

print("Total chunks collected:", len(all_chunks))
```

```python
all_chunks_path = folder / "all_chunks.json"

with open(all_chunks_path, "w", encoding="utf-8") as f:
    json.dump(all_chunks, f, indent=2)

print("Saved master chunk file:")
print(all_chunks_path)
```
