```python
from pydantic import BaseModel
from typing import Optional

class EvidenceFinding(BaseModel):
    study_title: Optional[str] = None
    year: Optional[int] = None
    population: Optional[str] = None
    sample_size: Optional[int] = None
    sepsis_type: Optional[str] = None
    biomarker: Optional[str] = None
    outcome: Optional[str] = None
    main_finding: Optional[str] = None
    odds_ratio: Optional[float] = None
    auc: Optional[float] = None
    p_value: Optional[str] = None
    source_quote: Optional[str] = None
    chunk_id: Optional[int] = None
```

```python
sample_ai_output = {
    "study_title": "Baloch et al.",
    "year": 2022,
    "population": "patients with sepsis",
    "sample_size": 120,
    "sepsis_type": "sepsis",
    "biomarker": "lactate",
    "outcome": "mortality",
    "main_finding": "Higher lactate was associated with increased mortality",
    "odds_ratio": 2.1,
    "auc": None,
    "p_value": "<0.05",
    "source_quote": "Higher lactate levels were associated with mortality.",
    "chunk_id": 0
}

validated_finding = EvidenceFinding(**sample_ai_output)

print(validated_finding)
```
